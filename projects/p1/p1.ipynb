### Adding Comments and Trying Our LLM Feedback System

#### Comments

In Python, you can add comments to your code by starting a line with `#`. Comments are pieces of text that Python ignores when it runs your program. They don’t change how your code behaves, but they are important because they let programmers:

* Explain what their code is doing.
* Leave reminders for themselves or others.
* Temporarily “turn off” a line of code without deleting it.

For example:

```Python
# This variable stores a greeting
msg = "Hello World"
```

#### LLM Feedback System and Using `##` to Ask Questions

In this course, we also have a **large language model (LLM) feedback system** integrated directly into the notebook, in which
- Regular comments starting with a single `#` are treated as normal comments for humans.
- Comments starting with `##` are treated as questions for the LLM feedback system.

When LLM feedback is enabled, any line beginning with `##` will be identified as a question and sent (along with your code) to our feedback server. You will then see additional explanations or guidance addressing your question alongside the usual test results.

The left side panel has icons for opening (or closing) the `File Browser`, the `Running Kernels and Terminals`, and the `Table of Contents` of your open notebook (as well as `extensions` which we won't go over).

#### Main Work Area

The main work area in the Jupyter Notebook interface enables you to arrange documents (notebooks, text files, etc.) and other activities (terminals, code consoles, etc.) into panels of tabs that can be resized or subdivided.  When you have a Jupyter notebook open you will have a bar of icons across the top of the tab:

<img src="images/top_icon_bar.png" alt="[IMG PATH BROKEN]"/> 

The icons on the left of the bar are for saving, adding a new cell, and cut, copy, and paste.  You then have a set of buttons for running the code in a cell, stopping the execution of code, restarting the kernel, and restarting the kernel and running all cells.  In the dropdown menu you can change the current cell type to either *Markdown*, *Code*, or *Raw*.  Python code goes into *Code* cells and text goes into *Markdown* cells.  We will not be using *Raw* cells.

On the far right of the bar are indicators to switch to *Notebook* mode (an older interface), open the *debugger* (which we will work with later), and to see the *status of the kernel*. When code is actively running, the circle in the far right of the top icon bar will become filled-in.  When no code is running the circle will be empty.

As mentioned above, a notebook is divided into cells.  The two cell types we will be using this semester are *Markdown* cells and *Code* cells.  There are several ways to interact with these cells.

The square brackets with a possible number inside indicate the order that cells are executed.  If an asterisk appears in the brackets then the cell is either currently being executed or about to be executed.  You can also click  and drag in this area to reorder the cells.  One common feature to turn on for the cells is line numbering.  If you click on the menu `View -> Show Line Numbers` then this side area will number the lines of the cell.
