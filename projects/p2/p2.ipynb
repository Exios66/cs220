### File systems, terminals and navigation on modern computers
In CS220, you will work like a programmer: you will navigate folders, create new directories, and run commands from a terminal. This gives you full control over where your files live and how your tools find them.

A terminal is a program that lets you interact with your computer by typing commands instead of clicking buttons. When you use a terminal, you give your computer instructions in text form, and the computer responds by running those instructions and showing the results.

Most of the time, you interact with your computer through a graphical interface—by clicking folders, dragging files, and opening apps. The terminal provides another way to do the same kinds of tasks, but using words instead of the mouse. For example, instead of clicking through folders to see what files are inside, you can type a command that lists them. Instead of right-clicking to create a folder, you can type a command to create it.

The terminal works by running a shell, which is a program that reads the commands you type and tells the operating system (Windows, macOS, Linux, etc) what to do. Common shell commands allow you to move between folders, see what files exist, create or delete files, and run other programs such as Python or Jupyter. Each command runs when you press Enter, and the terminal shows any output or error messages that result.

A **file system** is how your computer stores and organizes files. Think of it as a tree of folders (directories). The folder you are *currently in* is your **working directory**. The commands below can be run from a terminal to help you move around that tree and understand where you are.

#### Core navigation commands
- `pwd` (print working directory): shows the full path to your current folder

- `cd` (change directory): moves you to a different folder
- `mkdir` (make directory): creates a new folder, e.g., `mkdir labs`

#### Absolute vs. relative paths
A **path** is the address of a file or folder.

- **Absolute path**: starts from the root of the file system and fully specifies the location. Examples: `/Users/dummyUserName/Desktop/cs220-project-design` (macOS) or `C:\Users\dummyUserName\Desktop\cs220-project-design` (Windows).
- **Relative path**: starts from your current working directory. Examples: `p2` or `../p2` (one level up, then into `p2`).

#### Useful navigation shortcuts
- `.` means the current directory
- `..` means the parent directory (one level up)

### Segment 1: Introduction to Terminal & Interactive Mode (Python Shell)

When Python is run in **interactive mode**, it executes code **one line at a time** and immediately shows the result. Interactive mode runs when you start Python **directly from a terminal** by typing `pixi run python` (this is the default way to start Python in the pixi environment. When this happens, you will see a prompt like `>>>`, which means Python is waiting for you to type a line of code. This mode is useful for quick experiments, simple calculations, or checking how a small piece of code behaves.

A **Jupyter notebook is different from Python's interactive mode**, even though both allow you to run Python and see output right away. In a Jupyter notebook, Python does not run inside the terminal window where you typed the command to start Jupyter. Instead, Jupyter launches a **separate Python process in the background**, called a *kernel*. The notebook page you see in your browser is just an interface that sends code to this kernel.

When you click **Run** in a notebook, the entire cell is sent to the kernel and executed all at once. This is why Jupyter notebooks run code **cell by cell**, rather than one line at a time as you type. Although notebooks feel interactive, they are not the same as typing Python directly into a terminal.

The **terminal** is a separate tool used to run shell commands. Shell commands allow you to navigate your computer's file system, create and move files and folders, and start programs like Python or Jupyter. When you start Jupyter by running `pixi run jupyterlab`, you are doing so from a terminal. That terminal is now busy running the Jupyter server in the background.

Because the terminal that started Jupyter is occupied, you should **not use that same terminal window to run other commands**. Typing additional commands there could interrupt or stop Jupyter. Instead, you have two options for getting a new terminal to work in:

1. **Use the terminal built into JupyterLab**: go to **File → New → Terminal** from within JupyterLab. This terminal already has the pixi environment active, so you can run commands like `python` and `jupyter` directly.
2. **Open a new OS terminal tab or window**: on macOS, press **Command + T**; on Windows, press **Ctrl + Shift + N**; on Linux, press **Ctrl + Shift + T**. In this new terminal, run `pixi shell` first to activate the pixi environment before running any Python or project commands.

### Segment 4: Writing Boolean Expressions correctly

#### Background info and what NOT to do

You should always write boolean expressions in an expanded manner (examples below).
This is a **very important** lesson to learn. Make sure you go through this section very carefully, and let your TA/PM know if you have any difficulty with this section.

Consider the following sentence:

```
Is 3 + 4 equal to 6 or 7?
```

In English, you would say that this statement is True. However, writing it down in Python **exactly** as above would be **incorrect**. To see what happens, run the cell below:

````python
3 + 4 == 6 or 7
````

We **expect** to find that expression takes the value `True`, but we find that its value is `7`. 

In the above expression, what is the operator precedence? Please go back and refer to the precedence table.

As it turns out, the operator with the **highest precedence** here is `+`. So, in the expression above, the `3 + 4` gets evaluated first. So, the expression above is equivalent to the expression below:

````python
7 == 6 or 7
````

The operator with the **next highest precedence** here is `==`. Thus, `7 == 6` gets evaluated next. So, the expression above is equivalent to the expression below:

````python
False or 7
````

This is bad! You should **NEVER** use a boolean operator to **compare** a **non-boolean** value with a **boolean** value in this class!

To better understand what is going on here, try evaluating the following expressions:

````python
False or True
````

````python
False or False
````

````python
False or "hi"  # bad comparison
````

````python
False or 7  # bad comparison
````

**IMPORTANT LESSON:** Never use a boolean operator on non-boolean values in this class.

The **correct** way to express the English sentence

```
Is 3 + 4 equal to 6 or 7?
```

in Python is as follows:

````python
# this is the correct way to express the sentence above

3 + 4 == 6 or 3 + 4 == 7
````

### Segment 5: Modular Arithmetic

#### Clock example

An analog clock is a clock that displays time using a traditional clock face with moving hands.

What time does an analog clock show one hour after twelve o'clock? It wraps back around to 1 o'clock. This is a weird kind of arithmetic, where adding doesn't always make a number larger.

This alternative arithmetic is called **modular arithmetic**, and we can use the modulo operator (`%`) in Python to perform modular addition. The modulo operator works like division, but instead of giving you the *quotient*, it gives you the *remainder*. 

Below, you'll see many cells of the form `X%12`.  This computation is answering the question: if we start at 0 o'clock and wait X hours, what time is it? The `%12` part means time wraps around at 12 o'clock, meaning that there is no 12 o'clock, just 0 o'clock again (for the sake of this example, assume we have a "CS clock" whose hours go from 0 to 11 instead of 1 to 12).
